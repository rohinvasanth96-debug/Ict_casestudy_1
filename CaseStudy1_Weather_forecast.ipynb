```python
#Case study 2- Weather Forcast

temp=[29,26,33,35,27]
humidity=[70,71,75,73,71]
windspeed=[16.17,18,18,16]

temp_avg= sum(temp)/len(temp)
humidity_avg= sum(humidity)/len(humidity)
windspeed_avg= sum(windspeed)/len(windspeed)
print(f' Average temp= {temp_avg}, Average humidity= {humidity_avg}, Average windspeed= {windspeed_avg}')

if temp_avg>30 and humidity_avg>70 and windspeed_avg>17:
  print('Rainy')

elif temp_avg>35 and humidity_avg<60 and windspeed_avg<15:
  print('sunny')

else:
  print('cloudy')

```
