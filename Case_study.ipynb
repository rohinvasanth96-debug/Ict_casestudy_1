```python
#Case study 1-Bank Account

class Bank_account:

  def __init__(self,name,acno,bal):
    self.name=name
    self.acno=acno
    self.bal=bal

  #For withdrawal
  def withdraw (self,amt):
    self.amt=amt
    if self.amt>0:
      if self.amt<=self.bal:
        self.bal=self.bal-amt
        print(f' Amout withdrawn is {self.amt}. New balance is {self.bal}')
      else:
        print('insufficient balance')
    else:
      print("Invalid amount")

  #For deposit
  def deposit(self,amt):
    self.amt=amt
    if self.amt>0:
      self.bal=self.bal+self.amt
      print(f'Amount deposited is {self.amt}. New balace is {self.bal}')

    else:
      print("Invalid amount")

  #For balance enquiry
  def bal(bal):
    print(f'Your balance is {self.bal}')

#Creating objects

customer1=Bank_account('Rohin',101,1000)
customer2=Bank_account('Aswin',102,4000)

#Test
customer1.withdraw(100)
customer2.deposit(500)
print(customer1.bal)


```

```python
#Case study 2- Weather Forcast

temp=[29,26,33,35,27]
humidity=[70,71,75,73,71]
windspeed=[16.17,18,18,16]

temp_avg= sum(temp)/len(temp)
humidity_avg= sum(humidity)/len(humidity)
windspeed_avg= sum(windspeed)/len(windspeed)
print(f' Average temp= {temp_avg}, Average humidity= {humidity_avg}, Average windspeed= {windspeed_avg}')

if temp_avg>30 and humidity_avg>70 and windspeed_avg>17:
  print('Rainy')

elif temp_avg>35 and humidity_avg<60 and windspeed_avg<15:
  print('sunny')

else:
  print('cloudy')




```
